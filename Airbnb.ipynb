```python
import pandas as pd
import os
import shutil
```

```python
files = [f for f in os.listdir('/content') if f.endswith('.csv')]
files
```

```python
os.makedirs('/content/bronze', exist_ok=True)

for f in files:
    shutil.copy('/content/' + f, '/content/bronze/' + f)
```

```python
dfs = []

for f in files:
    df = pd.read_csv('/content/bronze/' + f)
    df['file_name'] = f
    dfs.append(df)

bronze_c = pd.concat(dfs, ignore_index=True)
bronze_c
```

```python
# dont need this column its only an index
silver.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# splitting file_name cuz it has city and day type together
silver['city'] = silver['file_name'].str.split('_').str[0]
silver['day_type'] = silver['file_name'].str.split('_').str[1].str.replace('.csv', '', regex=False)
silver[['file_name', 'city', 'day_type']].head()
```

```python
# realSum is the price for 2 nights this is what the owner of the data said so i divide it by 2 to get price per night
silver['price_per_night'] = silver['realSum'] / 2

# checking the result price per night should be half of realSum
silver[['realSum', 'price_per_night']].head()
```

```python
silver['is_outlier'] = silver['realSum'] > 3000

# checking how many outliers we marked
print(silver['is_outlier'].sum())
```

```python
from selenium import webdriver


options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument('--no-sandbox')

driver = webdriver.Chrome(options=options)
driver.get('https://en.wikipedia.org/wiki/List_of_European_cities_by_population_within_city_limits')
driver.title
```

```python
tables = pd.read_html(driver.page_source)
len(tables)
```

```python
tables[0].head()
```

```python
# first table is the one we need
cities = tables[0]

# wikipedia adds notes like [a] after names and numbers, so i cut the text
cities['City'] = cities['City'].str.split('[').str[0].str.lower()
cities['Official population'] = cities['Official population'].astype(str).str.split('[').str[0].str.replace(',', '')
cities['Official population'] = pd.to_numeric(cities['Official population'], errors='coerce')

cities.head()
```

```python

driver.quit()
```

```python
cities.shape
```

```python
# keep only the cities we have in silver
cities = cities[cities['City'].isin(silver['city'].unique())][['City', 'Country', 'Official population']]
print(cities.shape)

# my cities that are not in the wikipedia table
print(set(silver['city'].unique()) - set(cities['City']))
```

```python
# amsterdam, athens and lisbon are small so they are not in the wikipedia table, adding them
missing = pd.DataFrame({
    'City': ['amsterdam', 'athens', 'lisbon'],
    'Country': ['Netherlands', 'Greece', 'Portugal'],
    'Official population': [920000, 650000, 550000]
})

cities = pd.concat([cities, missing], ignore_index=True)
cities.shape
```

```python
# giving the columns simple names so they match silver
cities.columns = ['city', 'country', 'population']

# adding country and population to silver by matching the city name
silver = silver.merge(cities, on='city', how='left')

# checking every city should have a country and a population and rows stay 51707
print(silver.shape)
silver[['city', 'country', 'population']].drop_duplicates()
```

```python
# saving the final cleaned data as the silver master table
silver.to_csv('/content/silver_master.csv', index=False)
```

```python
os.listdir('/content')
```

```python
# gold keeps only normal listings not outliers
gold = silver[silver['is_outlier'] == False].reset_index(drop=True)

# the data has no id column so i make listing_id as the primary key
gold['listing_id'] = gold.index + 1

print(gold.shape)
gold[['listing_id', 'city', 'room_type', 'price_per_night']].head()
```

```python
#city_id is its primary key
dim_city = gold[['city', 'country', 'population']].drop_duplicates().reset_index(drop=True)
dim_city['city_id'] = dim_city.index + 1
dim_city
```

```python
# dimension_room_type and dimension_day_type, same idea as dim_city
dim_room_type = gold[['room_type']].drop_duplicates().reset_index(drop=True)
dim_room_type['room_type_id'] = dim_room_type.index + 1

dim_day_type = gold[['day_type']].drop_duplicates().reset_index(drop=True)
dim_day_type['day_type_id'] = dim_day_type.index + 1

print(dim_room_type)
print(dim_day_type)
```

```python
#replace the text columns with ids foreign keys
fact_listings = gold.merge(dim_city[['city', 'city_id']], on='city')
fact_listings = fact_listings.merge(dim_room_type, on='room_type')
fact_listings = fact_listings.merge(dim_day_type, on='day_type')
fact_listings = fact_listings.drop(columns=['city', 'country', 'population', 'room_type', 'day_type'])
fact_listings.shape
```

```python
# primary key must be unique
fact_listings['listing_id'].is_unique
```

```python
# the merge should not lose or duplicate any rows,
print(len(gold), len(fact_listings))
```

```python
# saving the tables for power bi
fact_listings.to_csv('/content/fact_listings.csv', index=False)
dim_city.to_csv('/content/dim_city.csv', index=False)
dim_room_type.to_csv('/content/dim_room_type.csv', index=False)
dim_day_type.to_csv('/content/dim_day_type.csv', index=False)
```
